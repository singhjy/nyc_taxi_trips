**Objective**: Download parquet files for yellow taxi trips.

**Source**: Web -  https://www.nyc.gov/site/tlc/about/tlc-trip-record-data.page

**Destination**: Local machine - save locally.

```python
import os
import urllib.request
```

```python
# 1. Define your specific folder path (Change this to your desired location)
output_folder = "/Users/jyotisingh/NY_Taxi_Trips"

# The web URL pointing to your Parquet file

#url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-01.parquet"

#url = "https://d37ci6vzurychx.cloudfront.net/trip-data/{color}_tripdata_{year}-{month}.parquet"

#Replace {color} (e.g., yellow, green, fhv), {year} (e.g., 2024), and {month} (e.g., 01) with your target values:

# 2. Setup download parameters
year = "2026"
color = "yellow"
months = ["01", "02", "03", "04", "05", "06"] 
base_url = "https://d37ci6vzurychx.cloudfront.net/trip-data"

# 3. Create the directory if it does not exist
os.makedirs(output_folder, exist_ok=True)

# 4. Loop and download
for month in months:
    filename = f"{color}_tripdata_{year}-{month}.parquet"
    url = f"{base_url}/{filename}"
    
    print(url)
    
    # Combine folder path and filename
    file_path = os.path.join(output_folder, filename)
    
    print(f"Downloading {filename} to {output_folder}...")
    try:
        urllib.request.urlretrieve(url, file_path)
        print(f"Successfully saved to {file_path}")
    except Exception as e:
        print(f"Failed to download {filename}: {e}")
```
